```python
import os
from PIL import Image
from pathlib import Path
import pandas as pd

project_images = [
    ('Proyecto 01', 'proyecto-01/01.jpg'),
    ('Proyecto 02', 'proyecto-02/01.jpg'),
]
rows = []
for name, rel in project_images:
    p = Path(rel)
    if p.exists():
        with Image.open(p) as img:
            rows.append({'proyecto': name, 'ruta': rel, 'ancho': img.width, 'alto': img.height})
    else:
        rows.append({'proyecto': name, 'ruta': rel, 'ancho': None, 'alto': None})

pd.DataFrame(rows)
```
